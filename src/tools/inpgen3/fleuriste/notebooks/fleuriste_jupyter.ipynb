# 🌱 FLEURiste — JupyterLab Interface

Interactive GUI for editing FLEUR `inp.xml` files, managing k-points,
generating input via inpgen, and creating SLURM job scripts — all from
within a Jupyter notebook.

## 1 — Launch the full GUI

The easiest way: call `fleuriste.jupyter()` which creates and displays
a tabbed interface with all four modes (XML Editor, K-Points, Inpgen,
Job Generator).

```python
import fleuriste

# Launch the full GUI — pass schema / input_file paths as needed
# gui = fleuriste.jupyter(schema='FleurInputSchema.xsd', input_file='inp.xml')
gui = fleuriste.jupyter()
```

## 2 — Using individual panels

You can also use each panel independently for embedding into your own
notebook workflow.

### 2a — XML Editor

```python
from fleuriste.jupyter_gui import XMLEditorPanel
from fleuriste.xml_editor import XMLDocument
from fleuriste.schema_parser import XSDParser

# Load schema and document
# schema = XSDParser('FleurInputSchema.xsd')
# doc = XMLDocument(schema=schema)
doc = XMLDocument()

# Load an inp.xml if available
from pathlib import Path
if Path('inp.xml').exists():
    doc.load('inp.xml')

editor = XMLEditorPanel(doc)
editor.widget
```

### 2b — K-Point Manager

```python
from fleuriste.jupyter_gui import KPointPanel

kpt = KPointPanel('inp.xml')
kpt.widget
```

### 2c — Inpgen (input generator)

```python
from fleuriste.jupyter_gui import InpgenPanel

inpgen = InpgenPanel()
inpgen.widget
```

### 2d — Job Generator (SLURM)

```python
from fleuriste.jupyter_gui import JobGeneratorPanel

job = JobGeneratorPanel('inp.xml')
job.widget
```
